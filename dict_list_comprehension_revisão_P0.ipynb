```python
# 1 Dada uma lista de números, crie uma nova lista apenas com os números negativos.

numeros = [1, -2, 3, -4, 5, -6]
negativos = [x for x in numeros if x < 0]
print (negativos)
```

```python
# 2 Dada uma lista de strings, crie uma lista só com as que têm mais de 5 caracteres.

strings = ["casa", "carro", "bicicleta", "avião", "ônibus"]
lista = [x for x in strings if len(x) > 5]
print (lista)
```

```python
# 3 Dada uma lista de números, crie uma lista com o cubo de cada um,
# mas só dos que são múltiplos de 3.

numeros = [1, 2, 3, 4, 5, 6]
cubos = [x**3 for x in numeros if x%3==0]
print (cubos)
```

```python
# 4 Dada uma frase, crie uma lista com o tamanho de cada palavra da frase.

frase = "O rato roeu a roupa do rei de Roma".split()
tamanhos = [len(x) for x in frase]
print (tamanhos)
```

```python
# 5 Dada uma string, use list comprehension para
# remover todas as vogais dela (gere a string sem vogais).

texto = "marina"
vogais = 'a e i o u'.split()
consoantes = ''.join([letra for letra in texto if letra not in vogais])
print(consoantes)
```

```python
# 6 Dada uma lista de nomes, ordene-a colocando primeiro os que começam com vogal, depois os demais
# (dica: a key pode devolver True/False).

nomes = ["Ana", "Bruno", "Carlos", "Daniela", "Eduardo"]
vogais = 'A E I O U'.split()
nomes_ordenados = sorted(nomes, key=lambda x: x[0] not in vogais)
print(nomes_ordenados)
```

```python
# 7 Dada uma lista de tuplas (nome, nota), ordene por nota, da maior pra menor.

alunos = [("Ana", 8.5), ("Bruno", 7.0), ("Carlos", 9.0), ("Daniela", 8.0), ("Eduardo", 9.5)]
ordenado = [sorted(alunos, key = lambda x: x[1], reverse=True)]
print(ordenado)
```

```python
# 8 Dada uma lista de palavras, ordene-a pela quantidade 
# de consoantes (não vogais) de cada palavra.

palavras = ["python", "programacao", "desenvolvimento", "computador"]
vogais = 'aeiou'.split()
ordenadas = sorted(palavras, key=lambda x: sum(1 for letra in x if letra not in vogais))
print(ordenadas)
```

```python
# 9 Dada uma lista de strings tipo ["item9", "item2", "item10"], ordene numericamente pelo número no final
# (não alfabeticamente — repare que "item10" não deve vir antes de "item2").

itens = ["item9", "item2", "item10"]
ordenados = sorted(itens, key=lambda x: int(x[4:]))
print(ordenados)
```

```python
# 10 Dada uma lista de palavras, ordene-a por ordem alfabética ignorando maiúsculas/minúsculas
# (ex: ["Ana", "bruno", "Carlos"]).

palavras = ["Carlos", "Ana", "bruno"]
ordenadas = sorted(palavras, key=lambda x: x.lower())
print(ordenadas)
```

```python
# 11 Dada uma lista de palavras, crie um dicionário 
# onde a chave é a palavra e o valor é o tamanho dela.

palavras = ["comprehension", "marina", "avião", "roxo"]
dicionario = {palavra: len(palavra) for palavra in palavras}
print(dicionario)
```

```python
# 12 Dado um dicionário de preços, crie um novo dicionário 
# só com os produtos que custam menos que um valor X.

precos = {"camisa": 100, "calça": 200, "sapato": 300, "vestido": 400}
dict = {produto: preco for produto, preco in precos.items() if preco < 250}
print(dict)
```

```python
# 13 Dada uma lista de números, crie um dicionário onde 
# a chave é o número e o valor é "par" ou "ímpar".

numeros = [1, 2, 3, 4, 5, 6]
dict = {numero: "par" if numero %2==0 else "ímpar" for numero in numeros}
print(dict)
```

```python
# 14 Dado um texto, crie um dicionário contando quantas vezes cada palavra aparece
# (não caractere — cuidado com maiúscula/minúscula e pontuação).
```

```python
# 15 Dado um dicionário {'joao': 25, 'maria': 30, 'ana': 22}, crie um novo dicionário
# só com quem tem idade acima da média do grupo.

pessoas = {'joao': 25, 'maria': 30, 'ana': 22}
dict = {nome: idade for nome, idade in pessoas.items() if idade>sum(pessoas.values())/len(pessoas)}
print(dict)
```

```python
# 16 Dada uma lista de palavras, crie um dicionário onde a chave é o tamanho da palavra
# e o valor é uma lista com todas as palavras daquele tamanho.

palavras = 'sol lua mar rio flor nuvem gato tigre'.split()
dicionario = {tamanho: [p for p in palavras if len(p) == tamanho]
              for tamanho in sorted(set(len(p) for p in palavras))}
print(dicionario)
```

```python
# 17 Dada uma lista de números, ordene-a colocando primeiro os pares (em ordem crescente)
# e depois os ímpares (em ordem crescente).

numeros = [1, 2, 3, 4, 5, 6]
lista = [x for x in sorted(numeros) if x % 2 == 0] + [x for x in sorted(numeros) if x % 2 != 0]
print(lista)
```

```python
# 18 Dado um dicionário {'produto': quantidade}, ordene-o do maior estoque pro menor,
# e depois pegue só os 3 primeiros.

estoque = {'produto1': 10, 'produto2': 5, 'produto3': 15, 'produto4': 8}
dict = {produto: quantidade for produto, quantidade in sorted(estoque.items(), key=lambda x: x[1], reverse=True)[:3]}
print(dict)
```

# Cópia vs referência em listas

```python
# 2 Crie uma lista a = [1, 2, 3]. Faça b = list(a) (cópia de verdade).
# Altere b[0] para 99. Preveja o que aparece ao imprimir a, e explique por quê.

a = [1, 2, 3]
b = list(a)
b[0] = 99
print(a)  # [1, 2, 3]
print(b)  # [99, 2, 3]
# Explicação: list(a) cria uma nova lista com os mesmos elementos de a. Portanto, alterar b não afeta a.
```

```python
# 3 Escreva uma função chamada zera_primeiro(lst) que recebe uma lista e muda
# o primeiro elemento dela para 0, sem usar return.
# Chame a função assim: numeros = [5, 6, 7]; zera_primeiro(numeros); print(numeros)
# Explique por que a lista original mudou mesmo sem return.

numeros = [5, 6, 7]
print(numeros)
def zera_primeiro(lst):
    lst[0] = 0
    # A lista original muda porque estamos modificando diretamente a referência para a lista, e não uma cópia dela.
zera_primeiro(numeros)
print(numeros)  # [0, 6, 7]
```

```python
# 5 Explique com suas palavras a diferença entre "==" e "is" quando comparamos
# duas listas em Python. Teste com: a = [1, 2, 3]; b = [1, 2, 3]; c = a
# Verifique a == b, a is b, a == c, a is c.

a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)  # True  → mesmo conteúdo
print(a is b)  # False → caixas diferentes na memória
print(a == c)  # True  → mesmo conteúdo (óbvio, é a mesma caixa)
print(a is c)  # True  → c é literalmente a mesma caixa que a, não uma cópia
```

# Dict comprehension com lista de dicionários

```python
# 6 Dada a lista de produtos abaixo, crie uma lista só com os nomes
# dos produtos que estão ativos (ativo == True).
produtos = [
    {'nome': 'Mouse', 'preco': 50, 'ativo': True},
    {'nome': 'Teclado', 'preco': 100, 'ativo': False},
    {'nome': 'Monitor', 'preco': 800, 'ativo': True}
]

ativos = [p['nome'] for p in produtos if p['ativo']]
print(ativos)  # ['Mouse', 'Monitor']
```

```python
# 7 Usando a mesma lista de produtos do exercício 6, crie um dicionário
# onde a chave é o nome do produto e o valor é o preço, mas apenas
# dos produtos ativos.

produtos = [
    {'nome': 'Mouse', 'preco': 50, 'ativo': True},
    {'nome': 'Teclado', 'preco': 100, 'ativo': False},
    {'nome': 'Monitor', 'preco': 800, 'ativo': True}
]

ativos = {p['nome']: p['preco'] for p in produtos if p['ativo']}
print(ativos)  # {'Mouse': 50, 'Monitor': 800}
```

```python
# 8 Usando a mesma lista de produtos, crie um novo dicionário {nome: preco}
# aplicando 10% de desconto no preço, apenas para os produtos com
# preco maior que 60.

produtos = [
    {'nome': 'Mouse', 'preco': 50, 'ativo': True},
    {'nome': 'Teclado', 'preco': 100, 'ativo': False},
    {'nome': 'Monitor', 'preco': 800, 'ativo': True}
]

desconto = {p['nome']: p['preco'] * 0.9 for p in produtos if p['preco'] > 60}
print(desconto)  # {'Teclado': 90.0, 'Monitor': 720.0}
```

```python
# 9 Dada a lista de alunos abaixo, crie um dicionário {nome: media}
# onde media é a média das notas de cada aluno.
alunos = [
    {'nome': 'Ana', 'notas': [8, 7, 9]},
    {'nome': 'Bruno', 'notas': [5, 6, 4]},
    {'nome': 'Carla', 'notas': [10, 9, 10]}
]

dicionario = {aluno['nome']: sum(aluno['notas']) / len(aluno['notas']) for aluno in alunos}
print(dicionario)  # {'Ana': 8.0, 'Bruno': 5.0, 'Carla': 9.666666666666666}
```

```python
# 10 Usando a lista de alunos do exercício 9, crie uma lista só com os
# nomes dos alunos cuja média é maior ou igual a 7.

alunos = [
    {'nome': 'Ana', 'notas': [8, 7, 9]},
    {'nome': 'Bruno', 'notas': [5, 6, 4]},
    {'nome': 'Carla', 'notas': [10, 9, 10]}
]

alunos_com_media_alta = [aluno['nome'] for aluno in alunos if sum(aluno['notas']) / len(aluno['notas']) >= 7]
print(alunos_com_media_alta)  # ['Ana', 'Carla']
```
