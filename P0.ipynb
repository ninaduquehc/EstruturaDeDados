```python
# 1 Dada uma lista de números, crie uma nova lista apenas com os números negativos.

numeros = [1, -2, 3, -4, 5, -6]
negativos = [x for x in numeros if x < 0]
print (negativos)
```

```python
# 2 Dada uma lista de strings, crie uma lista só com as que têm mais de 5 caracteres.

strings = ["casa", "carro", "bicicleta", "avião", "ônibus"]
lista = [x for x in strings if len(x) > 5]
print (lista)
```

```python
# 3 Dada uma lista de números, crie uma lista com o cubo de cada um,
# mas só dos que são múltiplos de 3.

numeros = [1, 2, 3, 4, 5, 6]
cubos = [x**3 for x in numeros if x%3==0]
print (cubos)
```

```python
# 4 Dada uma frase, crie uma lista com o tamanho de cada palavra da frase.

frase = "O rato roeu a roupa do rei de Roma".split()
tamanhos = [len(x) for x in frase]
print (tamanhos)
```

```python
# 5 Dada uma string, use list comprehension para
# remover todas as vogais dela (gere a string sem vogais).

texto = "marina"
vogais = 'a e i o u'.split()
consoantes = ''.join([letra for letra in texto if letra not in vogais])
print(consoantes)
```

```python
# 6 Dada uma lista de nomes, ordene-a colocando primeiro os que começam com vogal, depois os demais
# (dica: a key pode devolver True/False).

nomes = ["Ana", "Bruno", "Carlos", "Daniela", "Eduardo"]
vogais = 'A E I O U'.split()
nomes_ordenados = sorted(nomes, key=lambda x: x[0] not in vogais)
print(nomes_ordenados)
```

```python
# 7 Dada uma lista de tuplas (nome, nota), ordene por nota, da maior pra menor.

alunos = [("Ana", 8.5), ("Bruno", 7.0), ("Carlos", 9.0), ("Daniela", 8.0), ("Eduardo", 9.5)]
ordenado = [sorted(alunos, key = lambda x: x[1], reverse=True)]
print(ordenado)
```

```python
# 8 Dada uma lista de palavras, ordene-a pela quantidade 
# de consoantes (não vogais) de cada palavra.

palavras = ["python", "programacao", "desenvolvimento", "computador"]
vogais = 'aeiou'.split()
ordenadas = sorted(palavras, key=lambda x: sum(1 for letra in x if letra not in vogais))
print(ordenadas)
```

```python
# 9 Dada uma lista de strings tipo ["item9", "item2", "item10"], ordene numericamente pelo número no final
# (não alfabeticamente — repare que "item10" não deve vir antes de "item2").

itens = ["item9", "item2", "item10"]
ordenados = sorted(itens, key=lambda x: int(x[4:]))
print(ordenados)
```
