```python
#4 Dada uma lista de palavras, ordene-a pelo número de vogais presentes em cada palavra.

nomes = 'ana luiza maria felipe tiago'.split()
vogais = 'a e i o u'.split()
def letras(nome):
    soma = 0
    for letra in nome:
        if letra in vogais:
            soma += 1
    return soma
lista = [(letras(nome), nome) for nome in nomes]
lista.sort()
lista = [nome for quantidade, nome in lista]
print(lista)
```

```python
#5 Dada uma lista de palavras, ordene-a pelo último caractere de cada palavra.

nomes = 'tiago ana luiza maria felipe'.split()
lista = [(nome[-1], nome) for nome in nomes]
lista.sort()
nomes = [nome for caractere, nome in lista]
print(nomes)
```

```python
#6 Dada uma string, utilize list comprehension para criar uma nova string onde 
# os caracteres aparecem alternando entre maiúsculas e  minúsculas.

texto = "marina"
lista = [texto[i].upper() if i % 2 == 0 else texto[i].lower() for i in range(len(texto))]
resultado = ''.join(lista)
print(resultado)
```
