```python
# 1 Escreva uma função recursiva max_lista(lst) que retorna o maior
# elemento de uma lista. Não use a função max() pronta.
# Ex: max_lista([3, 8, 2, 5]) -> 8

def max_lista(lst):
    if len(lst) == 1:
        return lst[0]
    else:
        max_restante = max_lista(lst[1:])
        return lst[0] if lst[0] > max_restante else max_restante

print(max_lista([3, 8, 2, 5]))
```

```python
# 2 Escreva uma função recursiva pal(s) que verifica se uma string é
# um palíndromo (lê-se igual de trás pra frente).
# Ex: pal('radar') -> True | pal('python') -> False

def pal(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return pal(s[1:-1])

print(pal('radar'))  # True
print(pal('python'))  # False
```

```python
# 3 Escreva uma função recursiva conta_vogal(s) que conta quantas
# vogais existem em uma string.
# Ex: conta_vogal('abacate') -> 4

def conta_vogal(s):
    if len(s) == 0:
        return 0
    else:
        primeira_letra = s[0].lower()
        if primeira_letra in 'aeiou':
            return 1 + conta_vogal(s[1:])
        else:
            return conta_vogal(s[1:])

print(conta_vogal('abacate'))  # 4
```

```python
# 4 Escreva uma função recursiva soma_par(n) que soma todos os números
# pares de 1 até n (inclusive).
# Ex: soma_par(6) -> 2+4+6 -> 12

def soma_par(n):
    if n <= 1:
        return 0
    elif n % 2 == 0:
        return n + soma_par(n - 2)
    else:
        return soma_par(n - 1)

print(soma_par(6))  # 12
print(soma_par(9))  # 20
```

```python
# 5 Escreva uma função recursiva conta_ocorrencias(lst, valor) que conta
# quantas vezes um valor aparece dentro de uma lista.
# Ex: conta_ocorrencias([1, 2, 1, 3, 1], 1) -> 3

def conta_ocorrencias(lst, valor):
    if len(lst) == 0:
        return 0
    if lst[0] == valor:
        return 1 + conta_ocorrencias(lst[1:], valor)
    else:
        return 0 + conta_ocorrencias(lst[1:], valor)

print(conta_ocorrencias([1, 2, 1, 3, 1], 1))  # 3
print(conta_ocorrencias([1, 2, 3, 4, 5], 1))  # 1
```

```python
# 6 Escreva uma função recursiva eh_ordenada(lst) que verifica se uma
# lista de números está em ordem crescente.
# Ex: eh_ordenada([1, 3, 5, 8]) -> True | eh_ordenada([1, 5, 2]) -> False

def eh_ordenada(lst):
    if len(lst) <= 1:
        return True
    if lst[0] > lst[1]:
        return False
    return eh_ordenada(lst[1:])

print(eh_ordenada([1, 3, 5, 8]))  # True
print(eh_ordenada([1, 5, 2]))  # False
```

```python
# 7 Escreva uma função recursiva multiplica(a, b) que calcula a * b
# usando apenas somas (sem usar o operador *).
# Ex: multiplica(4, 3) -> 12

def multiplica(a, b):
    if b == 0:
        return 0
    else:
        return a + multiplica(a, b - 1)

print(multiplica(4, 3))  # 12
print(multiplica(5, 2))  # 10
```

```python
# 8 Escreva uma função recursiva conta_letra(s, letra) que conta quantas
# vezes uma letra específica aparece em uma string.
# Ex: conta_letra('banana', 'a') -> 3

def conta_letra(s, letra):
    if len(s) == 0:
        return 0
    if s[0] == letra:
        return 1 + conta_letra(s[1:], letra)
    else:
        return 0 + conta_letra(s[1:], letra)

print(conta_letra('banana', 'a'))  # 3
print(conta_letra('banana', 'b'))  # 1
```

```python
# 9 Escreva uma função recursiva potencia_soma(n) que calcula 1 + 2 + 4 + 8 + ...
# (soma de potências de 2) até 2^n.
# Ex: potencia_soma(3) -> 1+2+4+8 -> 15

def potencia_soma(n):
    if n < 0:
        return 0
    else:
        return (2 ** n) + potencia_soma(n - 1)

print(potencia_soma(3))  # 15
print(potencia_soma(4))  # 31
```

```python
# 10 Escreva uma função recursiva eh_palindromo_numero(n) que verifica se
# um número inteiro é um palíndromo (lê-se igual de trás pra frente),
# convertendo para string internamente.
# Ex: eh_palindromo_numero(1221) -> True | eh_palindromo_numero(123) -> False

def eh_palindromo_numero(n):
    s = str(n)
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return eh_palindromo_numero(s[1:-1])

print(eh_palindromo_numero(1221))  # True
print(eh_palindromo_numero(123))   # False
```
