```python
# 1) Dada uma lista de dicionários Python em que cada elemento possui as
# chaves 'nome', 'preco' (valor numérico) e 'ativo' (valor booleano),
# construa, obrigatoriamente usando dict comprehension, um novo
# dicionário que contenha apenas os produtos com 'ativo' igual a True,
# mapeando o nome do produto para o respectivo preço com acréscimo de
# 10%. Não é permitido utilizar laço for tradicional nem a função dict()
# com gerador. Exemplo: para a entrada [{'nome': 'Mouse', 'preco': 50,
# 'ativo': True}, {'nome': 'Teclado', 'preco': 100, 'ativo': False},
# {'nome': 'Monitor', 'preco': 300, 'ativo': True}], a saída esperada é
# {'Mouse': 55.0, 'Monitor': 330.0}. Para a entrada [{'nome': 'Camera',
# 'preco': 500, 'ativo': False}], a saída esperada é {}.

produtos = [
    {'nome': 'Mouse', 'preco': 50, 'ativo': True},
    {'nome': 'Teclado', 'preco': 100, 'ativo': False},
    {'nome': 'Monitor', 'preco': 300, 'ativo': True}
]
resultado = {p['nome']: p['preco'] * 1.10 for p in produtos if p['ativo']}
print(resultado)
```

```python
# 2) Dada uma lista de códigos alfanuméricos em que cada elemento contém
# exatamente um bloco contínuo de um ou mais dígitos, podendo aparecer
# em qualquer posição da string, escreva um código que ordene a lista em
# ordem crescente com base no valor numérico contido em cada string. A
# solução deve utilizar a função sorted() com o parâmetro key=. Exemplo:
# para a entrada ["item20A", "ref3B", "prod105X", "abc12Z"], a saída
# esperada é ["ref3B", "abc12Z", "item20A", "prod105X"].

codigos = ["item20A", "ref3B", "prod105X", "abc12Z"]
resultado = [x for x in sorted(codigos, key=lambda s: int(''.join([c for c in s if c in '0123456789'])))]
print(resultado)
```

```python
# 3) Dada a lista ['sol', 'estrela', 'lua', 'planeta'], ordene-a pelo
# número de consoantes de cada palavra. Considere apenas letras do
# alfabeto, trate maiúsculas e minúsculas como equivalentes e
# desconsidere acentos. Em caso de empate, mantenha a ordem original dos
# elementos. Faça em uma única linha utilizando key= e uma função
# anônima com lambda.

palavras = ['sol', 'estrela', 'lua', 'planeta']
vogais = 'aeiou'
resultado = sorted(palavras, key=lambda p: sum([1 for c in p.lower() if c not in vogais]))
print(resultado)
```

```python
# 4) Dada uma frase que não contém pontuação e cujas palavras estão
# separadas por um único espaço, construa um dicionário em que as chaves
# sejam as palavras distintas da frase (desconsiderando diferenças entre
# maiúsculas e minúsculas) e os valores sejam a quantidade de vezes que
# cada palavra aparece. Utilize dict comprehension. Exemplo: para a
# entrada "Python é simples e python é poderoso", a saída esperada é
# {'python': 2, 'é': 2, 'simples': 1, 'e': 1, 'poderoso': 1}.

palavras = "Python é simples e python é poderoso".lower().split()
resultado = {palavra: palavras.count(palavra) for palavra in palavras}
print(resultado)
```

```python
# 6) Escreva uma função chamada adjacente(n) que receba um número
# inteiro positivo n e retorne True caso existam dois dígitos vizinhos
# cuja diferença absoluta seja exatamente 1, seja crescente ou
# decrescente. Caso contrário, retorne False. Exemplos: adjacente(1219)
# -> True; adjacente(8642) -> False; adjacente(98) -> True; adjacente(55)
# -> False.

def adjacente(n):
    digitos = str(n)
    i = 0
    while i < len(digitos) - 1:
        atual = int(digitos[i])
        proximo = int(digitos[i + 1])
        if atual - proximo == 1 or proximo - atual == 1:
            return True
        i = i + 1
    return False


# Testes
print(adjacente(1219))  # True
print(adjacente(8642))  # False
print(adjacente(98))    # True
print(adjacente(55))    # False
```

```python
# 7) Escreva uma função chamada alarme(ativo, hora) em que ativo é um
# valor booleano indicando se o alarme está disparado e hora é um
# inteiro entre 0 e 23. Há problema se o alarme estiver ativo antes das
# 6 ou depois das 22. A função deve retornar "Problema: madrugada" se
# ativo for True e hora < 6; "Problema: noite" se ativo for True e hora
# > 22; e "Sem problemas" nos demais casos.

def alarme(ativo, hora):
    if ativo and hora < 6:
        return "Problema: madrugada"
    elif ativo and hora > 22:
        return "Problema: noite"
    else:
        return "Sem problemas"


# Testes
print(alarme(True, 3))    # Problema: madrugada
print(alarme(True, 23))   # Problema: noite
print(alarme(True, 12))   # Sem problemas
print(alarme(False, 3))   # Sem problemas
```

```python
# 8) Faça uma função soma_pares_finais(n) que receba um número inteiro
# positivo e retorne a soma dos dígitos pares consecutivos que aparecem
# no final do número. Considere apenas o bloco final de dígitos
# consecutivos pares, interrompendo ao encontrar o primeiro dígito
# ímpar. Exemplos: soma_pares_finais(2486) -> 14; soma_pares_finais(1358)
# -> 8; soma_pares_finais(7777) -> 0.

def soma_pares_finais(n):
    digitos = str(n)
    if len(digitos) < 2:
        return 0
    ultimo = int(digitos[-1])
    penultimo = int(digitos[-2])
    if ultimo % 2 == 0 and penultimo % 2 == 0:
        return ultimo + penultimo
    return 0


print(soma_pares_finais(2486))  # 14
print(soma_pares_finais(1358))  # 8 -> DEU ERROOOO
print(soma_pares_finais(7777))  # 0
```

```python
# 9) Dada uma frase, inverta cada palavra individualmente, mantendo a
# ordem original das palavras na frase. Considere que as palavras estão
# separadas por um único espaço. Escreva a função inverte(frase).
# Exemplos: inverte("ana e mariana gostam e banana") -> "ana e anairam
# matsog e ananab"; inverte("python é divertido") -> "nohtyp é
# odotrevnid".

def inverte(frase):
    palavras = frase.split()
    resultado = []
    i = 0
    while i < len(palavras):
        palavra_invertida = palavras[i][::-1]
        resultado.append(palavra_invertida)
        i = i + 1
    return ' '.join(resultado)


# Testes
print(inverte("ana e mariana gostam e banana"))
# ana e anairam matsog e ananab
print(inverte("python é divertido"))
# nohtyp é odotrevnid
```
