```python
# Calcule o fatorial de um número x.
def fat(x):
    if x == 0 or x == 1: return 1
    return x * fat(x - 1)
print(fat(5))
```

```python
# Calcule x elevado a y (potência), usando recursão.
def potencia(x,y):
    if y == 0: return 1
    return x*potencia(x,y-1)
print (potencia(2,3))
```

```python
# Inverta uma string usando recursão.
def inv(s):
    if len(s)==1: return s
    return inv(s[1:]) + s[0]
print (inv('abacate'))
```

```python
# Calcule a soma dos dígitos de um número inteiro n.
def sd(n):
    if n<=9: return n
    return n%10+sd(n//10)
print (sd(123))
```

```python
# Calcule o n-ésimo termo da sequência de Fibonacci (com memoização).
from functools import cache
@cache
def fib(n):
    if n==1 or n==2: return 1
    return fib(n-1)+fib(n-2)
print (fib(100))
```

```python
# Calcule o máximo divisor comum (MDC) entre a e b, usando o algoritmo de Euclides.
def mdc(a,b):
    if a%b==0: return b
    return mdc(b, a%b)
print (mdc(21, 15))
```

```python
# Converta um número decimal n para sua representação em binário (como string).
def bi(n):
    if n==0: return ''
    return bi(n//2)+str(n%2)
print (bi(18))
```

```python
# Calcule 2 elevado a n, usando recursão simples.
def f1(n):
    if n==0: return 1
    return 2 * f1(n-1)
print(f1(100))
```

```python
# Calcule 2 elevado a n, usando recursão dupla com memoização (f(n) = f(n-1) + f(n-1)).
from functools import cache
@cache
def f(n):
    if n==0: return 1
    return f(n-1) + f(n-1)
print(f(100))
```

```python
# Conte quantos dígitos tem o número n.
def cont_dig(n):
    n_str=str(n)
    if len(n_str)==1: return 1
    return 1 + cont_dig(n_str[1:])
print (cont_dig(12345))
```
